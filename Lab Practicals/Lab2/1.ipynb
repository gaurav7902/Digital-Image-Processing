```python
grayscale_img = cv2.imread("Images_Practice_Assignment/Camera_man.png", 0)

# 1 direct funtion
# rotated_img = cv2.rotate(grayscale_img, 0)

# 2 transpose then mirror
# rotated_img = grayscale_img
# N = rotated_img.shape[0]

# for i in range(N):
#     for j in range(i, N):
#         rotated_img[i][j], rotated_img[j][i] = rotated_img[j][i], rotated_img[i][j]

# for row in rotated_img:
#     for j in range(N // 2):
#         row[j], row[N - 1 - j] = row[N - 1 - j], row[j]

# 3 formula
rows = len(grayscale_img)
cols = len(grayscale_img[0])

rotated_img = [[0] * rows for _ in range(cols)]

for r in range(rows):
    for c in range(cols):   
        rotated_img[c][rows - 1 - r] = grayscale_img[r][c]

plt.imshow(rotated_img, cmap='gray')
plt.axis('off')
```

```python
# 1 direct funtion
# mirrored_img = [row[::-1] for row in grayscale_img]

# 2 for loop
mirrored_img = grayscale_img

for row in mirrored_img:
    n = len(row)
    for i in range(n // 2):
        row[i], row[n - 1 - i] = row[n - 1 - i], row[i]

plt.imshow(mirrored_img, cmap='gray')
plt.axis('off')
```

# Q3

```python
grayscale_img = cv2.imread("Images_Practice_Assignment/Natural_scene.jpg", 0)

y, x = 100, 150
# Select 8x8 patch
patch_8x8 = grayscale_img[y : y + 8, x : x + 8]
_, bin_img = cv2.threshold(patch_8x8, 50, 255, cv2.THRESH_BINARY)

plt.subplot(1,2,1)
plt.imshow(patch_8x8, cmap='gray')
plt.title('Patch')
plt.subplot(1,2,2)
plt.imshow(bin_img, cmap='gray')
plt.title('Binary Img')
```

```python
import numpy as np

bin_img = np.array([
    [0,255,255,0],
    [255,0,0,255],
    [0,0,255,0],
    [255,255,255,0]
])

N = bin_img.shape[0]
M = bin_img.shape[1]

subpart_2x2 = np.zeros((2,2), np.uint8)
for i in range(2):
    for j in range(2):
        subpart_2x2[i][j] = bin_img[i][j]

cnt = -1 # not considering first subpart
for i in range(0,N-1):
    for j in range(0,M-1):
        # if np.array_equal(bin_img[i:i+2,j:j+2],subpart_2x2):
        #     cnt+=1
        if (bin_img[i][j] == subpart_2x2[0][0] and bin_img[i][j+1]   == subpart_2x2[0][1] and
            bin_img[i+1][j] == subpart_2x2[1][0] and bin_img[i+1][j+1] == subpart_2x2[1][1]):
            cnt += 1
        
if(cnt>0):
    print(True, cnt)
else:
    print(False, 0)
```
